```python
# Diagnóstico de calidad de datos: La Liga
**Pregunta analítica:** ¿Qué factores influyen más en la posición final de los equipos?
```

```python
import pandas as pd

df_liga = pd.read_csv("../data/raw/la_liga.csv", encoding="utf-16")
print("Filas y columnas:", df_liga.shape)
df_liga.head()
```

```python
df_liga.head()
```

```python
df_liga.info()
```

```python
import numpy as np

# 1. ¿Score coincide con los goles?
score_calc = df_liga["Home Goals"].astype(str) + "-" + df_liga["Away Goals"].astype(str)
print("Score inconsistente:", (df_liga["Score"] != score_calc).sum())

# 2. ¿Teams coincide con Home Team - Away Team?
teams_calc = df_liga["Home Team"] + " - " + df_liga["Away Team"]
print("Teams inconsistente:", (df_liga["Teams"] != teams_calc).sum())

# 3. ¿Pools coincide con el resultado real?
pools_calc = np.where(df_liga["Home Goals"] > df_liga["Away Goals"], "1",
             np.where(df_liga["Home Goals"] < df_liga["Away Goals"], "2", "X"))
print("Pools inconsistente:", (df_liga["Pools"] != pools_calc).sum())

# 4. Valores de Pools
print(df_liga["Pools"].value_counts())
```

```python
# 6. Cobertura temporal y partidos por temporada
print("Temporadas:", df_liga["Season"].min(), "a", df_liga["Season"].max())
df_liga.groupby("Season").size().describe()
```

```python
# 7. Nombres de equipos (para detectar variantes del mismo club)
equipos = sorted(set(df_liga["Home Team"]) | set(df_liga["Away Team"]))
print(len(equipos))
equipos
```

```python
# 8. Partidos repetidos (misma temporada, jornada y equipos)
df_liga.duplicated(subset=["Season", "Game", "Home Team", "Away Team"]).sum()
```

```python
import numpy as np

print("=== FALTANTES ===")
print(df_liga.isnull().sum())

print("\n=== DUPLICADOS ===")
print("Filas totalmente duplicadas:", df_liga.duplicated().sum())
print("Partidos repetidos (temporada, jornada, equipos):",
      df_liga.duplicated(subset=["Season", "Game", "Home Team", "Away Team"]).sum())

print("\n=== CONSISTENCIA ===")
score_calc = df_liga["Home Goals"].astype(str) + "-" + df_liga["Away Goals"].astype(str)
print("Score inconsistente:", (df_liga["Score"] != score_calc).sum())

teams_calc = df_liga["Home Team"] + " - " + df_liga["Away Team"]
print("Teams inconsistente:", (df_liga["Teams"] != teams_calc).sum())

pools_calc = np.where(df_liga["Home Goals"] > df_liga["Away Goals"], "1",
             np.where(df_liga["Home Goals"] < df_liga["Away Goals"], "2", "X"))
print("Pools inconsistente:", (df_liga["Pools"] != pools_calc).sum())

print("\n=== PUNTOS POR VICTORIA POR ÉPOCA ===")
pts_ganador = df_liga[df_liga["Pools"] == "1"].groupby("Season")["Home Points"].max()
print("Primera temporada con 3 puntos:", pts_ganador[pts_ganador == 3].index.min())
print("Última temporada con 2 puntos:", pts_ganador[pts_ganador == 2].index.max())

print("\n=== PARTIDOS POR TEMPORADA ===")
print(df_liga.groupby("Season").size().describe())

print("\n=== EQUIPOS ===")
equipos = sorted(set(df_liga["Home Team"]) | set(df_liga["Away Team"]))
print(len(equipos))
print(equipos)
```

```python
# 1. Los 3 registros con Pools inconsistente
df_liga[df_liga["Pools"] != pools_calc]
```

```python
# 2. Temporadas faltantes
print("Faltan:", sorted(set(range(1929, 2024)) - set(df_liga["Season"])))
```

```python
# 3. ¿Los puntos coinciden con el resultado según la época?
gana_local = df_liga["Home Goals"] > df_liga["Away Goals"]
gana_visit = df_liga["Home Goals"] < df_liga["Away Goals"]
empate = df_liga["Home Goals"] == df_liga["Away Goals"]
pts_victoria = np.where(df_liga["Season"] >= 1996, 3, 2)

local_esp = np.where(gana_local, pts_victoria, np.where(empate, 1, 0))
visit_esp = np.where(gana_visit, pts_victoria, np.where(empate, 1, 0))
print("Home Points inconsistentes:", (df_liga["Home Points"] != local_esp).sum())
print("Away Points inconsistentes:", (df_liga["Away Points"] != visit_esp).sum())
```

```python
# 4. Temporadas en que aparece cada variante de nombre
variantes = ["Alavés", "Deportivo Alavés", "R. Sociedad", "Real Sociedad",
             "Almería", "Almeria AD", "Málaga", "CD Málaga", "Burgos", "Real Burgos CF"]
for v in variantes:
    s = pd.concat([df_liga.loc[df_liga["Home Team"] == v, "Season"],
                   df_liga.loc[df_liga["Away Team"] == v, "Season"]])
    print(f"{v:18} {s.min()}-{s.max()}   partidos: {len(s)}")
```
